```python
sns.set_theme(style="whitegrid")
plt.subplots(figsize=(7,5))
sns.scatterplot(data=tips, x='total_bill', y='tip', hue='time', style='smoker', size='size')
plt.legend(loc='upper left')
plt.title('Tips vs Total Bill')
plt.show()
```

```python
sns.boxplot(data=tips, x='day', y='total_bill', hue='smoker', palette='Accent')
plt.legend(loc='upper right')
plt.title('Total Bill by Day and Smoker Status')
plt.show()
```

```python
sns.violinplot(data=png, x='species', y='body_mass_g', hue='sex', split=True)
plt.title('Body Mass Distribution by Species and Sex')
plt.show()
```

```python
g = sns.pairplot(png, hue='species', corner=True, diag_kind='kde', palette='bright')
g.add_legend()
g.fig.suptitle('Penguins Pairwise Relationships')
plt.show()
```

```python
g = sns.relplot(data=tips, x='total_bill', y='tip', col='time', row='sex', hue='smoker')
g.fig.suptitle('Tips faceted by Time and Sex', y=1.03)
plt.show()
```

```python
g = sns.lmplot(data=tips, x='total_bill', y='tip', col='smoker', hue='smoker', palette='magma')
g.fig.suptitle('Tips vs Total Bill with Regression Line', y=1.05)
plt.show()
```

```python
g = sns.jointplot(data=png, x='flipper_length_mm', y='body_mass_g', hue='species')
g.fig.suptitle('Flipper Length vs Body Mass', y=1.05)
plt.show()
```

```python
sns.barplot(data=tips, x='day', y='total_bill', hue='sex', errorbar='sd', palette='Set2')
plt.title('Average Total Bill by Day and Sex (error = std dev)')
plt.show()
```

### 1. Qualitative Palettes (For Categorical Data)
Used when categories have no inherent order or magnitude (e.g., countries, genders, species).

* **Seaborn Defaults & Variations:** `'deep'` (default), `'muted'`, `'pastel'`, `'bright'`, `'dark'`, `'colorblind'`
* **ColorBrewer Palettes:** `'Set1'`, `'Set2'`, `'Set3'`, `'Paired'`, `'Dark2'`, `'Accent'`, `'Pastel1'`, `'Pastel2'`
* **Matplotlib Defaults:** `'tab10'`, `'tab20'`, `'tab20b'`, `'tab20c'`
* **Circular Systems (auto-generated hues):** `'hls'`, `'husl'`

---

### 2. Sequential Palettes (For Numeric / Gradients)
Used when data progresses continuously from low to high values (e.g., density, count, temperature).

* **Seaborn Native:** `'rocket'`, `'mako'`, `'flare'`, `'crest'`
* **Perceptually Uniform (Matplotlib):** `'viridis'`, `'plasma'`, `'inferno'`, `'magma'`, `'cividis'`
* **Single-Hue (Matplotlib):** `'Blues'`, `'Greens'`, `'Reds'`, `'Oranges'`, `'Purples'`, `'Greys'`
* **Multi-Hue (Matplotlib):** `'YlGnBu'`, `'YlOrRd'`, `'PuBuGn'`, `'GnBu'`, `'BuPu'`

---

### 3. Diverging Palettes (For Midpoint Data)
Used when data has a critical midpoint, like $0$ or a mean value, with variation extending in two directions (e.g., correlation matrices, profit/loss, temperature anomaly).

* **Seaborn Native:** `'vlag'`, `'icefire'`
* **Standard Diverging:** `'coolwarm'`, `'bwr'`, `'seismic'`, `'Spectral'`
* **ColorBrewer Diverging:** `'RdBu'`, `'RdGy'`, `'PiYG'`, `'PRGn'`, `'BrBG'`, `'PuOr'`, `'RdYlBu'`, `'RdYlGn'`
